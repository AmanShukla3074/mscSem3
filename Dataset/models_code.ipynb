```python
import bpy
import random
from mathutils import Vector

# ============================================================
# SETTINGS
# ============================================================

GRASS_PATH = r"C:\Projects\mscSem3\Dataset\models\3d_grass.glb"
NILGAI_PATH = r"C:\Projects\mscSem3\Dataset\models\nelgai\source\model.glb"

GRASS_SCALE = 1.2
GRASS_GROUPS = 12

# Grass vertical adjustment
GRASS_Z_OFFSET = -0.25

# Grass distribution area
GRASS_X_RANGE = 5
GRASS_Y_RANGE = 3

random.seed(42)

# ============================================================
# CLEAR SCENE
# ============================================================

bpy.ops.object.select_all(action='SELECT')
bpy.ops.object.delete(use_global=False)

# ============================================================
# HELPER FUNCTIONS
# ============================================================

def get_new_objects(before):
    return [
        obj for obj in bpy.context.scene.objects
        if obj not in before
    ]


def get_world_bounds(objects):

    bpy.context.view_layer.update()

    points = []

    for obj in objects:

        for corner in obj.bound_box:

            world_point = obj.matrix_world @ Vector(corner)

            points.append(world_point)

    min_x = min(p.x for p in points)
    max_x = max(p.x for p in points)

    min_y = min(p.y for p in points)
    max_y = max(p.y for p in points)

    min_z = min(p.z for p in points)
    max_z = max(p.z for p in points)

    return (
        min_x,
        max_x,
        min_y,
        max_y,
        min_z,
        max_z
    )


def move_objects_z(objects, amount):

    for obj in objects:

        matrix = obj.matrix_world.copy()

        matrix.translation.z += amount

        obj.matrix_world = matrix

    bpy.context.view_layer.update()


# ============================================================
# IMPORT GRASS
# ============================================================

before = set(bpy.context.scene.objects)

bpy.ops.import_scene.gltf(filepath=GRASS_PATH)

grass_objects = get_new_objects(before)

print("Grass objects imported:", len(grass_objects))


# ============================================================
# SCALE GRASS
# ============================================================

for obj in grass_objects:

    obj.scale *= GRASS_SCALE

bpy.context.view_layer.update()


# ============================================================
# GROUND ORIGINAL GRASS
# ============================================================

grass_bounds = get_world_bounds(grass_objects)

grass_min_z = grass_bounds[4]

move_objects_z(
    grass_objects,
    -grass_min_z + GRASS_Z_OFFSET
)

print("Original grass grounded.")


# ============================================================
# CREATE GRASS PATCHES
# ============================================================

original_grass = list(grass_objects)

for i in range(GRASS_GROUPS):

    # Random horizontal position
    x = random.uniform(
        -GRASS_X_RANGE,
        GRASS_X_RANGE
    )

    y = random.uniform(
        -GRASS_Y_RANGE,
        GRASS_Y_RANGE
    )

    # Random rotation
    rotation = random.uniform(
        0,
        6.283185
    )

    for obj in original_grass:

        # Create duplicate
        new_obj = obj.copy()

        # Copy mesh data
        if obj.data:

            new_obj.data = obj.data.copy()

        # Add duplicate to scene
        bpy.context.collection.objects.link(new_obj)

        # Preserve original world transform
        matrix = obj.matrix_world.copy()

        # Move only horizontally
        matrix.translation.x += x
        matrix.translation.y += y

        new_obj.matrix_world = matrix

        # Rotate around Z axis
        new_obj.rotation_euler[2] += rotation


print("Grass patches created.")


# ============================================================
# IMPORT NILGAI
# ============================================================

before = set(bpy.context.scene.objects)

bpy.ops.import_scene.gltf(filepath=NILGAI_PATH)

nilgai_objects = get_new_objects(before)

print("Nilgai objects imported:", len(nilgai_objects))


# ============================================================
# GET NILGAI BOUNDS
# ============================================================

nilgai_bounds = get_world_bounds(
    nilgai_objects
)

nilgai_min_x = nilgai_bounds[0]
nilgai_max_x = nilgai_bounds[1]

nilgai_min_y = nilgai_bounds[2]
nilgai_max_y = nilgai_bounds[3]

nilgai_min_z = nilgai_bounds[4]


# ============================================================
# CENTER NILGAI
# ============================================================

nilgai_center_x = (
    nilgai_min_x +
    nilgai_max_x
) / 2

nilgai_center_y = (
    nilgai_min_y +
    nilgai_max_y
) / 2


for obj in nilgai_objects:

    matrix = obj.matrix_world.copy()

    matrix.translation.x -= nilgai_center_x

    matrix.translation.y -= nilgai_center_y

    obj.matrix_world = matrix


bpy.context.view_layer.update()


# ============================================================
# GROUND NILGAI
# ============================================================

nilgai_bounds = get_world_bounds(
    nilgai_objects
)

nilgai_min_z = nilgai_bounds[4]

move_objects_z(
    nilgai_objects,
    -nilgai_min_z
)


# ============================================================
# FINAL CHECK
# ============================================================

grass_final_bounds = get_world_bounds(
    grass_objects
)

nilgai_final_bounds = get_world_bounds(
    nilgai_objects
)


print("--------------------------------------")
print("NILGAI FIELD READY")
print("--------------------------------------")

print("Grass scale:", GRASS_SCALE)
print("Grass groups:", GRASS_GROUPS)

print(
    "Grass X range:",
    -GRASS_X_RANGE,
    "to",
    GRASS_X_RANGE
)

print(
    "Grass Y range:",
    -GRASS_Y_RANGE,
    "to",
    GRASS_Y_RANGE
)

print(
    "Grass lowest Z:",
    grass_final_bounds[4]
)

print(
    "Nilgai lowest Z:",
    nilgai_final_bounds[4]
)

print("Nilgai centered at X=0, Y=0")

print("--------------------------------------")
```

```python
import bpy
import random
from mathutils import Vector

# ============================================================
# SETTINGS
# ============================================================

GRASS_PATH = r"C:\Projects\mscSem3\Dataset\models\3d_grass.glb"

NILGAI_PATH = r"C:\Projects\mscSem3\Dataset\models\nelgai\source\model.glb"


# ============================================================
# SOIL SETTINGS
# ============================================================

# Size of the soil field
SOIL_SIZE_X = 14
SOIL_SIZE_Y = 10

# Soil thickness
SOIL_THICKNESS = 0.15

# Small vertical position adjustment
SOIL_Z = -0.05


# ============================================================
# GRASS SETTINGS
# ============================================================

GRASS_SCALE = 1.2

GRASS_GROUPS = 12

# IMPORTANT:
# This is your calibrated value.
# DO NOT CHANGE for now.
GRASS_Z_OFFSET = -0.25

# Grass distribution area
GRASS_X_RANGE = 5
GRASS_Y_RANGE = 3


# ============================================================
# RANDOM SEED
# ============================================================

random.seed(42)


# ============================================================
# CLEAR SCENE
# ============================================================

bpy.ops.object.select_all(action='SELECT')
bpy.ops.object.delete(use_global=False)


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def get_new_objects(before):

    return [
        obj
        for obj in bpy.context.scene.objects
        if obj not in before
    ]


def get_world_bounds(objects):

    bpy.context.view_layer.update()

    points = []

    for obj in objects:

        for corner in obj.bound_box:

            world_point = obj.matrix_world @ Vector(corner)

            points.append(world_point)

    min_x = min(p.x for p in points)
    max_x = max(p.x for p in points)

    min_y = min(p.y for p in points)
    max_y = max(p.y for p in points)

    min_z = min(p.z for p in points)
    max_z = max(p.z for p in points)

    return (
        min_x,
        max_x,
        min_y,
        max_y,
        min_z,
        max_z
    )


def move_objects_z(objects, amount):

    for obj in objects:

        matrix = obj.matrix_world.copy()

        matrix.translation.z += amount

        obj.matrix_world = matrix

    bpy.context.view_layer.update()


# ============================================================
# CREATE SOIL
# ============================================================

print("--------------------------------------")
print("CREATING BLENDER SOIL")
print("--------------------------------------")


bpy.ops.mesh.primitive_cube_add(
    location=(0, 0, SOIL_Z)
)

soil = bpy.context.object

soil.name = "SOIL_GROUND"


# ============================================================
# SCALE SOIL
# ============================================================

soil.dimensions = (
    SOIL_SIZE_X,
    SOIL_SIZE_Y,
    SOIL_THICKNESS
)

bpy.ops.object.transform_apply(
    location=False,
    rotation=False,
    scale=True
)


# ============================================================
# SOIL MATERIAL
# ============================================================

soil_material = bpy.data.materials.new(
    name="Dry_Soil_Material"
)

soil_material.use_nodes = True

nodes = soil_material.node_tree.nodes
links = soil_material.node_tree.links

# Clear default nodes
nodes.clear()


# Output
output = nodes.new(
    "ShaderNodeOutputMaterial"
)

output.location = (500, 0)


# Principled BSDF
principled = nodes.new(
    "ShaderNodeBsdfPrincipled"
)

principled.location = (200, 0)


# Noise texture
noise = nodes.new(
    "ShaderNodeTexNoise"
)

noise.location = (-500, 100)

noise.inputs["Scale"].default_value = 5.0
noise.inputs["Detail"].default_value = 3.0
noise.inputs["Roughness"].default_value = 0.8


# Color ramp
ramp = nodes.new(
    "ShaderNodeValToRGB"
)

ramp.location = (-200, 100)


# Dark dry soil
ramp.color_ramp.elements[0].position = 0.25
ramp.color_ramp.elements[0].color = (
    0.10,
    0.055,
    0.025,
    1
)

# Light dry soil
ramp.color_ramp.elements[1].position = 0.75
ramp.color_ramp.elements[1].color = (
    0.32,
    0.18,
    0.07,
    1
)


# Connect nodes
links.new(
    noise.outputs["Fac"],
    ramp.inputs["Fac"]
)

links.new(
    ramp.outputs["Color"],
    principled.inputs["Base Color"]
)

links.new(
    ramp.outputs["Color"],
    principled.inputs["Roughness"]
)

links.new(
    principled.outputs["BSDF"],
    output.inputs["Surface"]
)


# Roughness
principled.inputs["Roughness"].default_value = 0.9


# Assign material
soil.data.materials.append(
    soil_material
)


print("Soil created.")


# ============================================================
# IMPORT GRASS
# ============================================================

print("--------------------------------------")
print("IMPORTING GRASS")
print("--------------------------------------")


before = set(bpy.context.scene.objects)

bpy.ops.import_scene.gltf(
    filepath=GRASS_PATH
)

grass_objects = get_new_objects(before)


print(
    "Grass objects imported:",
    len(grass_objects)
)


# ============================================================
# SCALE GRASS
# ============================================================

for obj in grass_objects:

    obj.scale *= GRASS_SCALE


bpy.context.view_layer.update()


# ============================================================
# GROUND ORIGINAL GRASS
# ============================================================

grass_bounds = get_world_bounds(
    grass_objects
)

grass_min_z = grass_bounds[4]


move_objects_z(
    grass_objects,
    -grass_min_z + GRASS_Z_OFFSET
)


bpy.context.view_layer.update()


print("Original grass grounded.")


# ============================================================
# CREATE GRASS PATCHES
# ============================================================

original_grass = list(grass_objects)


for i in range(GRASS_GROUPS):

    # Random horizontal position
    x = random.uniform(
        -GRASS_X_RANGE,
        GRASS_X_RANGE
    )

    y = random.uniform(
        -GRASS_Y_RANGE,
        GRASS_Y_RANGE
    )


    # Random rotation
    rotation = random.uniform(
        0,
        6.283185
    )


    for obj in original_grass:

        # Create duplicate
        new_obj = obj.copy()


        # Copy mesh data
        if obj.data:

            new_obj.data = obj.data.copy()


        # Add duplicate
        bpy.context.collection.objects.link(
            new_obj
        )


        # Preserve original world transform
        matrix = obj.matrix_world.copy()


        # Move horizontally
        matrix.translation.x += x
        matrix.translation.y += y


        new_obj.matrix_world = matrix


        # Random Z rotation
        new_obj.rotation_euler[2] += rotation


print("Grass patches created.")


# ============================================================
# IMPORT NILGAI
# ============================================================

print("--------------------------------------")
print("IMPORTING NILGAI")
print("--------------------------------------")


before = set(bpy.context.scene.objects)

bpy.ops.import_scene.gltf(
    filepath=NILGAI_PATH
)

nilgai_objects = get_new_objects(before)


print(
    "Nilgai objects imported:",
    len(nilgai_objects)
)


# ============================================================
# GET NILGAI BOUNDS
# ============================================================

nilgai_bounds = get_world_bounds(
    nilgai_objects
)


nilgai_min_x = nilgai_bounds[0]
nilgai_max_x = nilgai_bounds[1]

nilgai_min_y = nilgai_bounds[2]
nilgai_max_y = nilgai_bounds[3]


# ============================================================
# CENTER NILGAI
# ============================================================

nilgai_center_x = (
    nilgai_min_x +
    nilgai_max_x
) / 2


nilgai_center_y = (
    nilgai_min_y +
    nilgai_max_y
) / 2


for obj in nilgai_objects:

    matrix = obj.matrix_world.copy()

    matrix.translation.x -= nilgai_center_x
    matrix.translation.y -= nilgai_center_y

    obj.matrix_world = matrix


bpy.context.view_layer.update()


# ============================================================
# GROUND NILGAI
# ============================================================

nilgai_bounds = get_world_bounds(
    nilgai_objects
)

nilgai_min_z = nilgai_bounds[4]


move_objects_z(
    nilgai_objects,
    -nilgai_min_z
)


bpy.context.view_layer.update()


# ============================================================
# FINAL CHECK
# ============================================================

soil_final_bounds = get_world_bounds(
    [soil]
)

grass_final_bounds = get_world_bounds(
    grass_objects
)

nilgai_final_bounds = get_world_bounds(
    nilgai_objects
)


print("--------------------------------------")
print("NILGAI + GRASS + SOIL SCENE READY")
print("--------------------------------------")


print(
    "Soil size:",
    SOIL_SIZE_X,
    "x",
    SOIL_SIZE_Y
)


print(
    "Soil lowest Z:",
    soil_final_bounds[4]
)


print(
    "Soil highest Z:",
    soil_final_bounds[5]
)


print(
    "Grass scale:",
    GRASS_SCALE
)


print(
    "Grass groups:",
    GRASS_GROUPS
)


print(
    "Grass X range:",
    -GRASS_X_RANGE,
    "to",
    GRASS_X_RANGE
)


print(
    "Grass Y range:",
    -GRASS_Y_RANGE,
    "to",
    GRASS_Y_RANGE
)


print(
    "Grass lowest Z:",
    grass_final_bounds[4]
)


print(
    "Nilgai lowest Z:",
    nilgai_final_bounds[4]
)


print(
    "Nilgai centered at X=0, Y=0"
)


print("--------------------------------------")
print("DONE")
print("--------------------------------------")
```

```python
import bpy
import os
import math

from mathutils import Vector, Matrix


# ============================================================
# OUTPUT
# ============================================================

OUTPUT_DIR = r"C:\Projects\mscSem3\Dataset\thermal_demo"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ============================================================
# FIND OBJECTS
# ============================================================

soil = bpy.data.objects.get("SOIL_GROUND")
nilgai = bpy.data.objects.get("mesh_0")

grass_objects = []

for obj in bpy.context.scene.objects:

    if obj.name.startswith("Object_2") and obj.type == 'MESH':
        grass_objects.append(obj)


if soil is None:
    raise RuntimeError("SOIL_GROUND not found.")

if nilgai is None:
    raise RuntimeError("mesh_0 not found.")

if len(grass_objects) == 0:
    raise RuntimeError("No grass objects found.")


print("--------------------------------------")
print("OBJECTS FOUND")
print("Soil:", soil.name)
print("Nilgai:", nilgai.name)
print("Grass:", len(grass_objects))
print("--------------------------------------")


# ============================================================
# SAVE ORIGINAL NILGAI TRANSFORM
# ============================================================

original_matrix = nilgai.matrix_world.copy()


# ============================================================
# CREATE CAMERA
# ============================================================

camera = bpy.data.objects.get("THERMAL_DEMO_CAMERA")

if camera is None:

    camera_data = bpy.data.cameras.new(
        "THERMAL_DEMO_CAMERA_DATA"
    )

    camera = bpy.data.objects.new(
        "THERMAL_DEMO_CAMERA",
        camera_data
    )

    bpy.context.scene.collection.objects.link(camera)


scene = bpy.context.scene

scene.camera = camera


# ============================================================
# CREATE LIGHT
# ============================================================

light = bpy.data.objects.get("THERMAL_DEMO_LIGHT")

if light is None:

    light_data = bpy.data.lights.new(
        "THERMAL_DEMO_LIGHT_DATA",
        type="AREA"
    )

    light_data.energy = 1200
    light_data.size = 8

    light = bpy.data.objects.new(
        "THERMAL_DEMO_LIGHT",
        light_data
    )

    bpy.context.scene.collection.objects.link(light)


# ============================================================
# OBJECT CENTER
# ============================================================

def get_object_center(obj):

    corners = [
        obj.matrix_world @ Vector(corner)
        for corner in obj.bound_box
    ]

    return sum(corners, Vector()) / len(corners)


nilgai_center = get_object_center(nilgai)


# ============================================================
# POINT OBJECT TOWARD TARGET
# ============================================================

def point_at(obj, target):

    direction = Vector(target) - obj.location

    obj.rotation_euler = direction.to_track_quat(
        '-Z',
        'Y'
    ).to_euler()


# ============================================================
# CAMERA
# ============================================================

camera.location = (
    nilgai_center.x,
    nilgai_center.y - 12,
    nilgai_center.z + 4
)

camera.data.lens = 50

point_at(
    camera,
    nilgai_center
)

scene.camera = camera


# ============================================================
# LIGHT
# ============================================================

light.location = (
    nilgai_center.x,
    nilgai_center.y - 5,
    nilgai_center.z + 8
)

point_at(
    light,
    nilgai_center
)


# ============================================================
# RENDER SETTINGS
# ============================================================

# Your Blender version supports BLENDER_EEVEE
scene.render.engine = 'BLENDER_EEVEE'

scene.render.resolution_x = 640
scene.render.resolution_y = 480
scene.render.resolution_percentage = 100

scene.render.image_settings.file_format = 'PNG'

scene.render.film_transparent = False


# ============================================================
# MATERIAL CREATION
# ============================================================

def create_material(name, color):

    material = bpy.data.materials.get(name)

    if material is None:
        material = bpy.data.materials.new(name)

    material.use_nodes = True

    nodes = material.node_tree.nodes

    bsdf = nodes.get("Principled BSDF")

    if bsdf is not None:

        bsdf.inputs["Base Color"].default_value = (
            color[0],
            color[1],
            color[2],
            1.0
        )

        bsdf.inputs["Roughness"].default_value = 0.9

    return material


# ============================================================
# GRAYSCALE
# ============================================================

gray_soil = create_material(
    "THERMAL_GRAY_SOIL",
    (0.15, 0.15, 0.15)
)

gray_grass = create_material(
    "THERMAL_GRAY_GRASS",
    (0.30, 0.30, 0.30)
)

gray_animal = create_material(
    "THERMAL_GRAY_ANIMAL",
    (0.85, 0.85, 0.85)
)


# ============================================================
# IRONBOW RED
# ============================================================

iron_soil = create_material(
    "THERMAL_IRON_SOIL",
    (0.03, 0.005, 0.005)
)

iron_grass = create_material(
    "THERMAL_IRON_GRASS",
    (0.08, 0.005, 0.005)
)

iron_animal = create_material(
    "THERMAL_IRON_ANIMAL",
    (1.0, 0.03, 0.005)
)


# ============================================================
# BLUE / RED
# ============================================================

blue_soil = create_material(
    "THERMAL_BLUE_SOIL",
    (0.005, 0.01, 0.12)
)

blue_grass = create_material(
    "THERMAL_BLUE_GRASS",
    (0.01, 0.03, 0.30)
)

blue_animal = create_material(
    "THERMAL_BLUE_ANIMAL",
    (1.0, 0.03, 0.005)
)


# ============================================================
# WHITE HOT
# ============================================================

white_soil = create_material(
    "THERMAL_WHITE_SOIL",
    (0.05, 0.05, 0.05)
)

white_grass = create_material(
    "THERMAL_WHITE_GRASS",
    (0.18, 0.18, 0.18)
)

white_animal = create_material(
    "THERMAL_WHITE_ANIMAL",
    (1.0, 1.0, 1.0)
)


# ============================================================
# PALETTES
# ============================================================

THERMAL_PALETTES = {

    "grayscale": {
        "soil": gray_soil,
        "grass": gray_grass,
        "animal": gray_animal
    },

    "ironbow_red": {
        "soil": iron_soil,
        "grass": iron_grass,
        "animal": iron_animal
    },

    "blue_red": {
        "soil": blue_soil,
        "grass": blue_grass,
        "animal": blue_animal
    },

    "white_hot": {
        "soil": white_soil,
        "grass": white_grass,
        "animal": white_animal
    }
}


# ============================================================
# ANIMAL POSITIONS
# ============================================================

ANIMAL_CONFIGS = [

    {
        "name": "front_center",
        "x": 0.0,
        "y": 0.0,
        "rotation": 0
    },

    {
        "name": "side_left",
        "x": -2.0,
        "y": 0.0,
        "rotation": 90
    },

    {
        "name": "angled_right",
        "x": 2.0,
        "y": 0.5,
        "rotation": 45
    },

    {
        "name": "rear_offset",
        "x": 0.5,
        "y": 1.5,
        "rotation": 180
    }
]


# ============================================================
# ASSIGN MATERIAL
# ============================================================

def assign_material(objects, material):

    for obj in objects:

        if obj.type != 'MESH':
            continue

        if len(obj.data.materials) == 0:

            obj.data.materials.append(material)

        else:

            obj.data.materials[0] = material


# ============================================================
# SET NILGAI POSE
# ============================================================

def set_animal_pose(config):

    base_matrix = original_matrix.copy()

    angle = math.radians(
        config["rotation"]
    )

    rotation_matrix = Matrix.Rotation(
        angle,
        4,
        'Z'
    )

    original_location = base_matrix.translation.copy()

    rotated_location = (
        rotation_matrix @ original_location
    )

    rotated_location.x += config["x"]
    rotated_location.y += config["y"]

    new_matrix = (
        rotation_matrix @ base_matrix
    )

    new_matrix.translation = rotated_location

    nilgai.matrix_world = new_matrix

    bpy.context.view_layer.update()


# ============================================================
# RENDER
# ============================================================

def render_thermal_image(
    palette_name,
    palette,
    config
):

    print(
        "Rendering:",
        palette_name,
        config["name"]
    )


    # Apply materials

    assign_material(
        [soil],
        palette["soil"]
    )

    assign_material(
        grass_objects,
        palette["grass"]
    )

    assign_material(
        [nilgai],
        palette["animal"]
    )


    # Move animal

    set_animal_pose(config)


    # Update camera target

    center = get_object_center(nilgai)

    point_at(
        camera,
        center
    )


    # Filename

    filename = (
        palette_name
        + "_"
        + config["name"]
        + ".png"
    )


    output_path = os.path.join(
        OUTPUT_DIR,
        filename
    )


    scene.render.filepath = output_path


    # Make absolutely sure camera is active

    scene.camera = camera


    # Render

    bpy.ops.render.render(
        write_still=True
    )


    print(
        "Generated:",
        filename
    )


# ============================================================
# GENERATE ALL 16 IMAGES
# ============================================================

print("")
print("======================================")
print("STARTING THERMAL DEMO GENERATION")
print("======================================")


image_count = 0


for palette_name, palette in THERMAL_PALETTES.items():

    for config in ANIMAL_CONFIGS:

        render_thermal_image(
            palette_name,
            palette,
            config
        )

        image_count += 1


# ============================================================
# RESTORE ORIGINAL NILGAI
# ============================================================

nilgai.matrix_world = original_matrix

bpy.context.view_layer.update()


# ============================================================
# COMPLETE
# ============================================================

print("")
print("======================================")
print("THERMAL DEMO COMPLETED")
print("======================================")
print("Total images:", image_count)
print("Output:", OUTPUT_DIR)
print("======================================")
```

```python
import bpy
import os
import math

# ============================================================
# DATASET SETTINGS
# ============================================================

OUTPUT_FOLDER = os.path.join(
    bpy.path.abspath("//"),
    "dataset_grayscale_ironbow"
)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# ============================================================
# FIND OBJECTS FROM YOUR SCENE
# ============================================================

soil = bpy.data.objects.get("SOIL_GROUND")
nilgai = bpy.data.objects.get("mesh_0")

grass_objects = []

for obj in bpy.context.scene.objects:
    if obj.type == 'MESH' and obj.name.startswith("Object_2"):
        grass_objects.append(obj)

print("\n========================================")
print("THERMAL DATASET GENERATOR")
print("========================================")

print("Soil:", soil.name if soil else "NOT FOUND")
print("Nilgai:", nilgai.name if nilgai else "NOT FOUND")
print("Grass objects:", len(grass_objects))

if soil is None:
    raise RuntimeError("SOIL_GROUND not found.")

if nilgai is None:
    raise RuntimeError("mesh_0 not found.")

if len(grass_objects) == 0:
    print("WARNING: No grass objects found.")

# ============================================================
# CREATE MATERIAL
# ============================================================

def create_material(name, color, roughness=0.8):

    mat = bpy.data.materials.get(name)

    if mat is None:
        mat = bpy.data.materials.new(name)

    mat.use_nodes = True

    nodes = mat.node_tree.nodes
    bsdf = nodes.get("Principled BSDF")

    if bsdf:
        bsdf.inputs["Base Color"].default_value = (
            color[0],
            color[1],
            color[2],
            1
        )

        bsdf.inputs["Roughness"].default_value = roughness

        # Keep material bright and visible
        if "Emission Color" in bsdf.inputs:
            bsdf.inputs["Emission Color"].default_value = (
                color[0],
                color[1],
                color[2],
                1
            )

        if "Emission Strength" in bsdf.inputs:
            bsdf.inputs["Emission Strength"].default_value = 0.15

    return mat


# ============================================================
# MATERIALS
# ============================================================

# ------------------------------------------------------------
# GRAYSCALE / WHITE-HOT
# ------------------------------------------------------------

gray_animal = create_material(
    "THERMAL_GRAY_ANIMAL",
    (0.95, 0.95, 0.95)
)

gray_grass = create_material(
    "THERMAL_GRAY_GRASS",
    (0.22, 0.22, 0.22)
)

gray_soil = create_material(
    "THERMAL_GRAY_SOIL",
    (0.10, 0.10, 0.10)
)

# ------------------------------------------------------------
# IRONBOW / GAME-LIKE FALSE COLOR
# ------------------------------------------------------------

ironbow_animal = create_material(
    "THERMAL_IRONBOW_ANIMAL",
    (1.0, 0.10, 0.01)
)

ironbow_grass = create_material(
    "THERMAL_IRONBOW_GRASS",
    (0.01, 0.02, 0.20)
)

ironbow_soil = create_material(
    "THERMAL_IRONBOW_SOIL",
    (0.02, 0.01, 0.08)
)

# ============================================================
# ASSIGN MATERIAL
# ============================================================

def assign_material(obj, material):

    if obj is None:
        return

    if len(obj.data.materials) == 0:
        obj.data.materials.append(material)
    else:
        obj.data.materials[0] = material


# ============================================================
# CAMERA
# ============================================================

camera = bpy.data.objects.get("THERMAL_FIXED_CAMERA")

if camera is None:

    camera_data = bpy.data.cameras.new("THERMAL_FIXED_CAMERA")

    camera = bpy.data.objects.new(
        "THERMAL_FIXED_CAMERA",
        camera_data
    )

    bpy.context.collection.objects.link(camera)

    print("Created fixed camera.")

# ------------------------------------------------------------
# FIXED CAMERA POSITION
# ------------------------------------------------------------

# IMPORTANT:
# This camera NEVER moves between images.

camera.location = (7.5, -10.5, 4.5)

camera.rotation_euler = (
    math.radians(72),
    0,
    math.radians(35)
)

camera.data.lens = 52

bpy.context.scene.camera = camera

print("Fixed camera:", camera.location)

# ============================================================
# LIGHT
# ============================================================

light = bpy.data.objects.get("THERMAL_FIXED_LIGHT")

if light is None:

    light_data = bpy.data.lights.new(
        "THERMAL_FIXED_LIGHT",
        type='AREA'
    )

    light = bpy.data.objects.new(
        "THERMAL_FIXED_LIGHT",
        light_data
    )

    bpy.context.collection.objects.link(light)

light.location = (4, -4, 8)

light.data.energy = 500
light.data.shape = 'DISK'
light.data.size = 8

# ============================================================
# WORLD BACKGROUND
# ============================================================

world = bpy.context.scene.world

if world is None:
    world = bpy.data.worlds.new("ThermalWorld")
    bpy.context.scene.world = world

world.use_nodes = True

background = world.node_tree.nodes.get("Background")

if background:
    background.inputs["Color"].default_value = (
        0.005,
        0.005,
        0.005,
        1
    )

    background.inputs["Strength"].default_value = 0.15

# ============================================================
# RENDER SETTINGS
# ============================================================

scene = bpy.context.scene

scene.render.engine = 'BLENDER_EEVEE'

scene.render.resolution_x = 640
scene.render.resolution_y = 480
scene.render.resolution_percentage = 100

scene.render.image_settings.file_format = 'PNG'

scene.render.film_transparent = False

# ============================================================
# MATERIAL SETUP
# ============================================================

def set_grayscale():

    print("Applying GRAYSCALE / WHITE-HOT materials")

    assign_material(nilgai, gray_animal)

    assign_material(soil, gray_soil)

    for obj in grass_objects:
        assign_material(obj, gray_grass)


def set_ironbow():

    print("Applying IRONBOW materials")

    assign_material(nilgai, ironbow_animal)

    assign_material(soil, ironbow_soil)

    for obj in grass_objects:
        assign_material(obj, ironbow_grass)


# ============================================================
# SAVE RENDER
# ============================================================

def render_image(filename):

    filepath = os.path.join(
        OUTPUT_FOLDER,
        filename
    )

    scene.render.filepath = filepath

    print("Rendering:", filename)

    bpy.ops.render.render(
        write_still=True
    )

    print("Saved:", filepath)


# ============================================================
# STORE ORIGINAL NILGAI TRANSFORM
# ============================================================

original_location = nilgai.location.copy()
original_rotation = nilgai.rotation_euler.copy()
original_scale = nilgai.scale.copy()

# ============================================================
# ANIMAL POSITIONS
# ============================================================

# The camera stays FIXED.
#
# Only the animal changes orientation.
#
# Left / right means the animal's head/body orientation changes
# while the camera remains in exactly the same place.

poses = {

    "left_1":  math.radians(20),

    "left_2":  math.radians(45),

    "right_1": math.radians(-20),

    "right_2": math.radians(-45)

}

# ============================================================
# GENERATE GRAYSCALE DATASET
# ============================================================

print("\n========================================")
print("GENERATING GRAYSCALE DATASET")
print("========================================")

set_grayscale()

for pose_name, rotation_z in poses.items():

    # Reset location
    nilgai.location = original_location.copy()

    # Change ONLY animal orientation
    nilgai.rotation_euler = original_rotation.copy()

    nilgai.rotation_euler.z = (
        original_rotation.z + rotation_z
    )

    render_image(
        "grayscale_" + pose_name + ".png"
    )

# ============================================================
# GENERATE IRONBOW DATASET
# ============================================================

print("\n========================================")
print("GENERATING IRONBOW DATASET")
print("========================================")

set_ironbow()

for pose_name, rotation_z in poses.items():

    # Reset location
    nilgai.location = original_location.copy()

    # Change ONLY animal orientation
    nilgai.rotation_euler = original_rotation.copy()

    nilgai.rotation_euler.z = (
        original_rotation.z + rotation_z
    )

    render_image(
        "ironbow_" + pose_name + ".png"
    )

# ============================================================
# RESTORE ORIGINAL MODEL
# ============================================================

nilgai.location = original_location
nilgai.rotation_euler = original_rotation
nilgai.scale = original_scale

print("\n========================================")
print("DATASET GENERATION COMPLETE")
print("========================================")

print("Output folder:")
print(OUTPUT_FOLDER)

print("\nGenerated files:")

for pose_name in poses.keys():

    print(
        "grayscale_" +
        pose_name +
        ".png"
    )

for pose_name in poses.keys():

    print(
        "ironbow_" +
        pose_name +
        ".png"
    )

print("========================================")
```
